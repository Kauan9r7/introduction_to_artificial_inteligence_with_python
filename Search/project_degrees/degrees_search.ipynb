### Imports

```python
import time
import random
from IPython.display import HTML, display

# Importa as estruturas e funções do próprio arquivo do projeto
from degrees import load_data, people, movies, shortest_path, person_id_for_name
```

### Base de dados

```python
# Escolha a pasta de dados: 'small' ou 'large'
DATASET_DIR = "data/large"

print(f"Carregando dados da pasta '{DATASET_DIR}'...")
start_time = time.time()

# Carrega os dicionários na memória
load_data(DATASET_DIR)

elapsed = time.time() - start_time
print(
    f"✅ Dados carregados em {elapsed:.2f}s! ({len(people)} pessoas, {len(movies)} filmes)"
)
```

### Função Auxiliar para Exibição

```python
def find_and_display_degrees(source_name, target_name,pick_first_name=False):
    """Busca o caminho mais curto entre dois atores e exibe um relatório formatado."""
    source_id = person_id_for_name(source_name,pick_first_name)
    target_id = person_id_for_name(target_name,pick_first_name)

    if source_id is None:
        print(f"❌ Ator/Atriz '{source_name}' não foi encontrado(a).")
        return
    if target_id is None:
        print(f"❌ Ator/Atriz '{target_name}' não foi encontrado(a).")
        return

    # Executa a busca em largura
    start_time = time.time()
    path = shortest_path(source_id, target_id)
    search_time = (time.time() - start_time) * 1000

    if path is None:
        display(
            HTML(
                f"<h3 style='color: #D32F2F;'>Nenhuma conexão encontrada entre {source_name} e {target_name}.</h3>"
            )
        )
        return

    degrees_count = len(path)

    # Monta a tabela HTML formatada com cores explícitas para garantir visibilidade
    html_content = f"""
    <div style="border: 2px solid #4CAF50; border-radius: 10px; padding: 15px; margin: 10px 0; background-color: #ffffff; color: #222222;">
        <h2 style="color: #2E7D32; margin-top: 0;">🎬 Conexão Encontrada!</h2>
        <p style="color: #333333;"><b>Origem:</b> {people[source_id]['name']} | <b>Destino:</b> {people[target_id]['name']}</p>
        <p style="color: #333333;"><b>Graus de Separação:</b> <span style="font-size: 18px; font-weight: bold; color: #1565C0;">{degrees_count}</span> | <b>Tempo de busca:</b> {search_time:.2f} ms</p>
        <hr style="border: 0.5px solid #ccc;">
        <table style="width:100%; border-collapse: collapse; font-family: sans-serif; color: #222222;">
            <thead>
                <tr style="background-color: #4CAF50; color: white;">
                    <th style="padding: 10px; text-align: center;">Passo</th>
                    <th style="padding: 10px; text-align: left;">Ator 1</th>
                    <th style="padding: 10px; text-align: left;">Filme Compartilhado</th>
                    <th style="padding: 10px; text-align: left;">Ator 2</th>
                </tr>
            </thead>
            <tbody>
    """

    person1 = source_id
    for i, (movie_id, person2) in enumerate(path, 1):
        p1_name = people[person1]["name"]
        p2_name = people[person2]["name"]
        m_title = movies[movie_id]["title"]
        m_year = movies[movie_id]["year"]

        bg_color = "#ffffff" if i % 2 != 0 else "#f4f4f6"
        html_content += f"""
                <tr style="background-color: {bg_color}; border-bottom: 1px solid #ddd;">
                    <td style="padding: 10px; text-align: center; font-weight: bold; color: #333333;">{i}</td>
                    <td style="padding: 10px; font-weight: bold; color: #111111;">{p1_name}</td>
                    <td style="padding: 10px; color: #B71C1C; font-weight: bold;">🎬 {m_title} ({m_year})</td>
                    <td style="padding: 10px; font-weight: bold; color: #111111;">{p2_name}</td>
                </tr>
        """
        person1 = person2

    html_content += """
            </tbody>
        </table>
    </div>
    """

    display(HTML(html_content))
def test_random_degrees():
    """Sorteia dois atores aleatórios da base de dados e busca a conexão entre eles."""
    if not people:
        print(
            "❌ Base de dados não carregada! Execute a célula de carregamento primeiro."
        )
        return

    # Garante a escolha de 2 pessoas distintas da base de dados
    person1_id, person2_id = random.sample(list(people.keys()), 2)

    name1 = people[person1_id]["name"]
    name2 = people[person2_id]["name"]

    print(f"🎲 Sorteados: {name1} ➔ {name2}\n")

    # Chama a função de busca e exibição formatada
    find_and_display_degrees(name1, name2,True)
```
