## Node and frotier

Define how the frontier and node work

```python
class Node:
    """Representa um nó na árvore de busca."""

    def __init__(self, state, parent=None, action=None, cost=0, heuristic=0):
        self.state = state
        self.parent = parent
        self.action = action
        self.cost = cost          # g(n) -> Custo percorrido
        self.heuristic = heuristic  # h(n) -> Estimativa até o objetivo

class search_algorithm:

    def __init__(self):
        self.frontier = []

    def add(self, node):
        self.frontier.append(node)

    def contains_state(self, state):
        return any(node.state == state for node in self.frontier)

    def empty(self):
        return len(self.frontier) == 0

    def remove(self):
        # Function remove define the algorithm
        return NotImplementedError
```

# Depth-First Search

```python
class DFS(search_algorithm):
    # Define the function that removes a node from the frontier and returns it.
    def remove(self):
        # Terminate the search if the frontier is empty, because this means that there is no solution.
        if self.empty():
            raise Exception("empty frontier")
        else:
            # Save the last item in the list (which is the newest node added)
            node = self.frontier[-1]
            # Save all the items on the list besides the last node (i.e. removing the last node)
            self.frontier = self.frontier[:-1]
            return node
```

# Breadth-First Search

```python
class BFS(search_algorithm):   
    # Define the function that removes a node from the frontier and returns it.
    def remove(self):
        # Terminate the search if the frontier is empty, because this means that there is no solution.
        if self.empty():
            raise Exception("empty frontier")
        else:
            # Save the oldest item on the list (which was the first one to be added)
            node = self.frontier[0]
            # Save all the items on the list besides the first one (i.e. removing the first node)
            self.frontier = self.frontier[1:]
            return node
```

# Greedy Best-First Search

```python
class GBFS(search_algorithm):
    def remove(self):
        if self.empty():
            raise Exception("empty frontier")
        else:
            # Encontra o nó na fronteira que possui o menor valor de heurística h(n)
            # Assumindo que a heurística está guardada em 'node.heuristic' 
            # ou calculada previamente ao adicionar o nó na fronteira.
            node = min(self.frontier, key=lambda n: n.heuristic)
            
            # Remove esse nó específico da fronteira
            self.frontier.remove(node)
            
            return node
```

# A* Search

```python
class ASTAR(search_algorithm):
    def remove(self):
        if self.empty():
            raise Exception("empty frontier")
        else:
            # No A*, selecionamos o nó com o menor valor de f(n) = g(n) + h(n)
            # Assumindo que o nó possui 'cost' g(n) e 'heuristic' h(n)
            node = min(self.frontier, key=lambda n: n.cost + n.heuristic)
            
            # Remove esse nó específico da fronteira
            self.frontier.remove(node)
            
            return node
```
