# Minimax

Explora toda a árvore de decisão de forma recursiva até atingir os nós terminais.

```python
def minimax(board):
    """
    Returns the optimal action for the current player on the board.
    Minimax padrão.
    """
    if terminal(board):
        return None

    current_player = player(board)

    if current_player == 0:
        best_val = -float("inf")
        best_action = None

        for action in actions(board):
            val = min_value(result(board, action))
            if val > best_val:
                best_val = val
                best_action = action
        return best_action

    else:
        best_val = float("inf")
        best_action = None

        for action in actions(board):
            val = max_value(result(board, action))
            if val < best_val:
                best_val = val
                best_action = action
        return best_action


def max_value(board):
    if terminal(board):
        return utility(board)

    v = -float("inf")
    for action in actions(board):
        v = max(v, min_value(result(board, action)))
    return v


def min_value(board):
    if terminal(board):
        return utility(board)

    v = float("inf")
    for action in actions(board):
        v = min(v, max_value(result(board, action)))
    return v
```

# Alpha-Beta Pruning

Otimização do Minimax que mantém o mesmo resultado final, mas ignora/corta ramos da árvore de busca que comprovadamente não alterarão a escolha do jogador.

## Funções do Jogo Necessárias

Para implementar o Alpha-Beta Pruning, o seu jogo precisa fornecer exatamente as mesmas funções do Minimax.

```python
def minimax_alpha_beta(board):
    """
    Returns the optimal action for the current player on the board.
    Uses Alpha-Beta Pruning for optimization.
    """
    if terminal(board):
        return None

    current_player = player(board)

    if current_player == 0:
        best_val = -float("inf")
        best_action = None
        alpha = -float("inf")
        beta = float("inf")

        for action in actions(board):
            val = min_value(result(board, action), alpha, beta)
            if val > best_val:
                best_val = val
                best_action = action
            alpha = max(alpha, best_val)
        return best_action

    else:
        best_val = float("inf")
        best_action = None
        alpha = -float("inf")
        beta = float("inf")

        for action in actions(board):
            val = max_value(result(board, action), alpha, beta)
            if val < best_val:
                best_val = val
                best_action = action
            beta = min(beta, best_val)
        return best_action


def max_value(board, alpha, beta):
    if terminal(board):
        return utility(board)

    v = -float("inf")
    for action in actions(board):
        v = max(v, min_value(result(board, action), alpha, beta))
        alpha = max(alpha, v)
        if beta <= alpha:
            break
    return v


def min_value(board, alpha, beta):
    if terminal(board):
        return utility(board)

    v = float("inf")
    for action in actions(board):
        v = min(v, max_value(result(board, action), alpha, beta))
        beta = min(beta, v)
        if beta <= alpha:
            break
    return v
```

# Depth-Limited Minimax

Interrompe a busca recursiva ao atingir um limite máximo de profundidade definido ($d$), em vez de ir obrigatoriamente até o fim do jogo.

```python
def depth_limited_minimax(board, max_depth):
    """
    Retorna a melhor ação para o jogador atual respeitando o limite de profundidade.
    """
    if terminal(board):
        return None

    current_player = player(board)

    if current_player == 0:
        best_val = -float("inf")
        best_action = None
        alpha = -float("inf")
        beta = float("inf")

        for action in actions(board):
            val = min_value(result(board, action), max_depth - 1, alpha, beta)
            if val > best_val:
                best_val = val
                best_action = action
            alpha = max(alpha, best_val)
        return best_action

    else:
        best_val = float("inf")
        best_action = None
        alpha = -float("inf")
        beta = float("inf")

        for action in actions(board):
            val = max_value(result(board, action), max_depth - 1, alpha, beta)
            if val < best_val:
                best_val = val
                best_action = action
            beta = min(beta, best_val)
        return best_action


def max_value(board, depth, alpha, beta):
    if terminal(board):
        return utility(board)
    if depth == 0:
        return evaluate(board)

    v = -float("inf")
    for action in actions(board):
        v = max(v, min_value(result(board, action), depth - 1, alpha, beta))
        alpha = max(alpha, v)
        if beta <= alpha:
            break
    return v


def min_value(board, depth, alpha, beta):
    if terminal(board):
        return utility(board)
    if depth == 0:
        return evaluate(board)

    v = float("inf")
    for action in actions(board):
        v = min(v, max_value(result(board, action), depth - 1, alpha, beta))
        beta = min(beta, v)
        if beta <= alpha:
            break
    return v
```
